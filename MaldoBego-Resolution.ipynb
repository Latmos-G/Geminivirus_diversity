```python
# Import the required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sb
sb.set(font_scale = 2.5)
```

```python
# Load the dataset and display the overview of first five rows

df = pd.read_csv('MaldoBego.csv')  # This read out the dataset
df.head()
```

```python
# Explore if either week days or weekend has discrepancy in the success of the payment transaction

plt.figure(figsize=(35, 27))  # Chosen size of the plot

# Plot the count of the transaction (both failed and succ)in the week day

count = sb.barplot(x = 'Virus accessions within Maldovirus and Begomovirus', y = 'Percentage similarity to KM386645 (AGV1) based on pairwise identity', 
                   hue = 'Selected virus components', data=df, palette = ['sandybrown', 'sienna', 'brown']
                  )
plt.legend(title = 'Selected virus components', fontsize = 30)

count.set_ylabel('Percentage (%) similarity to KM386645 (AGV1) based on pairwise identity',weight = 'bold', size = 30)
count.set_xlabel('Virus accessions within Maldovirus and Begomovirus')
```
